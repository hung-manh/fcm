```python
from pyspark.sql import SparkSession
from pyspark.ml.feature import VectorAssembler
from pyspark.sql.functions import col
from pyspark.sql.types import FloatType
import numpy as np
from models.fcm import Dfcm
```

```python
# Initialize Spark session
spark = SparkSession.builder.appName("FCM_PySpark").getOrCreate()
spark
```

```python
# Step 1: Load data from HDFS
csv_file_path ="data/csv/602_Dry_Bean.csv"
data = spark.read.csv(csv_file_path, header=True, inferSchema=True)

## Drop the last column (labels) 
data = data.drop(data.columns[-1])

## Convert all columns to float type    
for column in data.columns:
    data = data.withColumn(column, col(column).cast(FloatType()))

# Step 2: Prepare data for clustering
assembler = VectorAssembler(inputCols=data.columns, outputCol="features")
data_vectorized = assembler.transform(data)

# Convert to RDD
data_rdd = data_vectorized.select("features").rdd.map(lambda row: np.array(row["features"]))
```

```python
import numpy as np
from utils.utils import norm_distances

class Dfcm:
    def __init__(self, m: float = 2, epsilon: float = 1e-5, maxiter: int = 10000):
        self._m = m  
        self._epsilon = epsilon  
        self._maxiter = maxiter  
    
    @staticmethod
    def __init_membership(N: int, C: int, seed: int = 0) -> np.ndarray:
        if seed > 0:
            np.random.seed(seed)
        U0 = np.random.rand(N, C)
        return U0 / U0.sum(axis=1)[:, None]
    
    def update_cluster_centers(self, data , membership):
        um = membership ** self._m 
        return (um.T @ data) / um.sum(axis=0)[:, np.newaxis] 
    
    def update_membership_matrix(self, distances_rdd):
        def update_membership(distances):
            epsilon = 1e-10  
            distances = np.maximum(distances, epsilon)  
            U = distances[:, :, None] * (1 / distances)[:, None, :]
            U = (U ** (2 / (self._m - 1))).sum(axis=2)
            return 1 / U    
        return distances_rdd.map(update_membership) 
    
    def cmeans(self, data_rdd, C:int = 3, seed: int = 42) -> tuple:
        u = self.__init_membership(data_rdd.count(), C, seed)
        u_rdd = data_rdd.sparkContext.parallelize(u)

        for step in range(self._maxiter):
            old_u = u_rdd.collect()
            v = self.update_cluster_centers(data_rdd, old_u)

            sdistances_rdd = norm_distances(data, v) 

            u_rdd = self.update_membership_matrix(sdistances_rdd)
            
            if (np.abs(u - old_u)).max(axis=(0, 1)) < self._epsilon:
                break
        return u, v, step + 1
```

```python
# Step 3: Initialize the FCM model
fcm_model = Dfcm(m=2, epsilon=1e-5, maxiter=10000)

# Step 4: Run FCM algorithm
def run_fcm_partition(partition, C, seed):
    """
    Function to run FCM on a partition.
    """
    partition_data = np.array(list(partition))
    if len(partition_data) == 0:
        return iter([])
    u, v, step = fcm_model.cmeans(partition_data, C, seed)
    return iter([(u, v, step)])

# Number of clusters
C = 7
seed = 42

# Run FCM on RDD partitions
results_rdd = data_rdd.mapPartitions(lambda partition: run_fcm_partition(partition, C, seed))

# Collect results
results = results_rdd.collect()
print(results)

# Combine results from all partitions (This is a simplified step; further merging logic might be needed)
U = np.vstack([result[0] for result in results])
V = np.mean([result[1] for result in results], axis=0)  # Averaging centroids
iterations = max([result[2] for result in results])

print(f"Total iterations: {iterations}")
```
